```python
import pandas as pd
import matplotlib.pyplot as plt
import mplcyberpunk
plt.style.use("cyberpunk")
```

```python
url = (f"http://api.bcb.gov.br/dados/serie/bcdata.sgs.4189/dados?formato=json")
df = pd.read_json(url)
df["data"] = pd.to_datetime(df["data"], dayfirst=True).dt.date
df.set_index("data", inplace=True)
selic_220_period = df.tail(220)
selic_12_period = df.tail(12)
selic_hoje = df.tail(1)
plt.figure(figsize=(23, 4))
plt.subplot(1, 2, 1)
plt.plot(selic_220_period.index, selic_220_period["valor"],  label="Selic")
plt.title("Taxa Selic")
plt.subplot(1, 2, 2)
plt.plot(selic_12_period.index, selic_12_period["valor"], label="Selic")
plt.title("Taxa Selic")
plt.annotate(
    f"{selic_hoje['valor'].values[0]}%",
    xy=(selic_hoje.index[0], selic_hoje["valor"].values[0]),
    xytext=(selic_hoje.index[0], selic_hoje["valor"].values[0] + 1),
    arrowprops=dict(facecolor="blue", shrink=0.05),
)
plt.show()
```
